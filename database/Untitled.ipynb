```python
customers_data = customers[['customer_id', 'state']]
sns.histplot(data=customers_data, x='state')
customers_data['state'].value_counts()
```

```python
orders_and_clients = orders.merge(customers_data, on='customer_id', how='inner')
full_data = orders_and_clients.merge(order_items, on='order_id', how='inner')
df_plot = pd.DataFrame(full_data.groupby('state')['list_price'].sum()).reset_index()
sns.barplot(df_plot, x='state', y='list_price')
plt.ticklabel_format(style='plain', axis='y')
```

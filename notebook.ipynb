```python
# Устанавливаем сид для воспроизводимости
np.random.seed(42)

# DataFrame 1 — пользователи
df1 = pd.DataFrame({
    'user_id': np.arange(1, 51),
    'age': np.random.randint(18, 60, size=50),
    'city': np.random.choice(['Warsaw', 'Krakow', 'Gdansk', 'Wroclaw'], size=50),
    'signup_date': pd.to_datetime('2022-01-01') + pd.to_timedelta(np.random.randint(0, 365, size=50), unit='D')
})

# DataFrame 2 — покупки
df2 = pd.DataFrame({
    'user_id': np.random.choice(np.arange(1, 51), size=50, replace=True),
    'purchase_amount': np.round(np.random.uniform(10.0, 500.0, size=50), 2),
    'purchase_date': pd.to_datetime('2023-01-01') + pd.to_timedelta(np.random.randint(0, 365, size=50), unit='D'),
    'category': np.random.choice(['Books', 'Electronics', 'Clothing', 'Groceries'], size=50)
})
```

```python
merged_df
```

```python
avg_amount_per_person = merged_df[['user_id','purchase_amount', 'city']].groupby(['user_id','city']).mean().reset_index()
avg_amount_per_person
```

```python
df = avg_amount_per_person[['city', 'purchase_amount']].groupby('city').mean()
```

```python
df.nlargest(3, columns='purchase_amount')
```

```python
unique_categories_by_cities = merged_df.groupby('city')['category'].nunique().reset_index()
unique_categories_by_cities
```

```python
result = unique_categories_by_cities.merge(df, on='city')
result
```

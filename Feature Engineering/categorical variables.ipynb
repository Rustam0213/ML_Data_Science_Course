```python
my_object_df = df.select_dtypes(include='object')
my_numeric_df = df.select_dtypes(exclude='object')
```

```python
my_numeric_df.head()
```

```python
my_object_df.head()
```

```python
df_object_dummies = pd.get_dummies(my_object_df, drop_first=True, dtype=int)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('dm_office_sales.csv')
```

```python
df.head()
```

```python
df['division'].value_counts()
```

```python
plt.figure(figsize=(10,4), dpi=200)
sns.countplot(data=df, x='level of education', hue='division')
```

```python
plt.figure(figsize=(10,4))
sns.barplot(data=df, x='level of education', y='salary', hue='division', estimator=np.mean, errorbar='sd')

plt.legend(loc=(1.05,0.5))
# errorbar - параметр отвечающий за среднеквадратическое отклонение всех значений, наример средняя зарплата у 200 работников это 80к, 
# но есть те работники которые зарабатывают больше или меньше, это и даёт некое отклонение в значениях
```

```python
df = pd.read_csv('StudentsPerformance.csv')
df.head()
```

```python
plt.figure(dpi=200, figsize=(10,4))
sns.boxplot(data=df, y='reading score', x='parental level of education', hue='test preparation course')
```

```python
plt.figure(dpi=200, figsize=(10,6))
sns.violinplot(data=df, x='reading score', y='parental level of education', hue='parental level of education')
```

```python
plt.figure(dpi=200, figsize=(10,6))
sns.violinplot(data=df, x='reading score', y='parental level of education', hue='test preparation course', 
               split=True, inner='quartile', bw_adjust=0.3)
```

```python
plt.figure(figsize=(10,6), dpi=200)
sns.swarmplot(data=df, y='gender', x='math score', size=2, hue='test preparation course', dodge=True)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('dm_office_sales.csv')
```

```python
df.head()
```

```python
# plt.figure(figsize=(5,8), dpi=200)
sns.rugplot(x='salary', data=df, height=0.02)
```

```python
# sns.set(style='whitegrid')
sns.displot(data=df, x='salary', bins=20, color='red', edgecolor='blue', linewidth=2,
            kde=True, rug=True)
# sns.distplot - не используем
```

```python
sns.histplot(data=df, x='salary')
```

```python
sns.kdeplot(data=df, x='salary')
```

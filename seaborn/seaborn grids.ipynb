```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('StudentsPerformance.csv')
df.head()
```

```python
sns.catplot(data=df, x='gender', y='math score', kind='box', hue='gender', row='lunch', col='test preparation course')
```

```python
sns.catplot(data=df, x='gender', y='math score', kind='box', hue='gender', col='lunch', row='race/ethnicity')
```

```python
sns.catplot(data=df, x='gender', y='math score', kind='violin', hue='gender', col='lunch', row='race/ethnicity')
```

```python
g = sns.PairGrid(data=df, hue='gender')
g = g.map_upper(sns.scatterplot)
g = g.map_diag(sns.kdeplot)
g = g.map_lower(sns.kdeplot)
g = g.add_legend()
```

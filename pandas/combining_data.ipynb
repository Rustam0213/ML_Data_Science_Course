```python
import pandas as pd
import numpy as np
```

```python
data_one = {'A': ['A0', 'A1', 'A2', 'A3'],'B': ['B0', 'B1', 'B2', 'B3']}
data_two = {'C': ['C0', 'C1', 'C2', 'C3'], 'D': ['D0', 'D1', 'D2', 'D3']}
```

```python
one = pd.DataFrame(data_one)
```

```python
two = pd.DataFrame(data_two)
```

```python
one
```

```python
two
```

```python
pd.concat([one,two],axis=1) # соеденяет два дф в один по колонкам
```

```python
pd.concat([two,one],axis=1) # переворачиваем значения
```

```python
pd.concat([one,two],axis=0) # произойдёт такая белеберда из-за того что у нас имеются разные названия колонок,
# чтобы это обойти проще всего будет просто переименовать колонки со второго дф
```

```python
two.columns = one.columns
```

```python
two
```

```python
#теперь можем их объеденить по строкам
pd.concat([one,two], axis=0)
```

```python
registrations = pd.DataFrame({'reg_id':[1,2,3,4],'name':['Andrew','Bobo','Claire','David']})
logins = pd.DataFrame({'log_id':[1,2,3,4],'name':['Xavier','Andrew','Yolanda','Bobo']})
```

```python
registrations
```

```python
logins
```

```python
pd.merge(registrations, logins, how='inner', on='name')
```

```python
pd.merge(logins, registrations, how='inner', on='name')
```

```python
pd.merge(registrations, logins, how='left', on='name')
```

```python
pd.merge(registrations, logins, how='right', on='name')
```

```python
pd.merge(left = registrations, right = logins, how='right', on='name')
```

```python
pd.merge(registrations, logins, how='outer', on='name')
```

```python
registrations = registrations.set_index('name')
```

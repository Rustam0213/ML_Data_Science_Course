Сейчас будем искать последние 4 цифры номера счёта, но обратим внимание, что это число, числа нельзя перебирать в виде списка, по этому
нужно с помощью нашей функции переделать это число в строку а потом найти то что нам нужно.

```python
def last_four(num):
    return str(num)[-4:] # принимает автоматически данные в виде аргумента и соответственно преобразовует так как нам это нужно
```

```python
last_four(33435534345345)
```

```python
df['CC Number'].apply(last_four) # тем самым мы передаём нашу функцию в качестве аргумента, и она обрабатывает каждую запись в таблице
```

```python
df['last_four'] = df['CC Number'].apply(last_four)
```

```python
def yelp(price):
    if price < 10:
        return '$'
    elif price >= 10 and price < 30:
        return '$$'
    else:
        return '$$$'
```

```python
df['yelp'] = df['total_bill'].apply(yelp)
```

```python
def quality(total_bill, tip):
    if tip/total_bill >= 0.25:
        return "Щедрые чаевые"
    else:
        return 'Обчные чаевые'
```

```python
quality(100, 25)
```

```python
df['quality'] = df[['total_bill', 'tip']].apply(lambda df: quality(df['total_bill'], df['tip']), axis=1)
```

```python
df.drop('quality', axis=1)
```

```python
df['quality'] = np.vectorize(quality)(df['total_bill'], df['tip']) # запоминаем этот синтаксис так как он проще
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = np.linspace(0,10,11)
b = a ** 4
```

```python
x = np.arange(0,10)
y = 2 * x
```

```python
fig, axes_arr = plt.subplots(nrows=1, ncols=2)
axes_arr[0].plot(x,y)
axes_arr[1].plot(a,b)
```

```python
fig, axes_arr = plt.subplots(nrows=2, ncols=2) # каждый рядок это новый список, по этому ссылаемся соответственно по индексу
axes_arr[0][0].plot(x,y)
# axes_arr[0][1].plot()
axes_arr[1][1].plot(a,b)
```

```python
fig, axes_arr = plt.subplots(nrows=3, ncols=1)
for ax in axes_arr:
    ax.plot(x,y)

plt.tight_layout() # делаем расстояние чтобы графики не пересекались
```

```python
fig, axes_arr = plt.subplots(nrows=2, ncols=2)
axes_arr[0][0].plot(x,y)
axes_arr[0][1].plot(x,y)
axes_arr[1][0].plot(x,y)
axes_arr[1][1].plot(a,b)

fig.subplots_adjust(wspace=1, hspace=1) # тоже меняем расстояние между графиками используя параметры
```

```python
fig, axes_arr = plt.subplots(nrows=2, ncols=2)

axes_arr[0][0].plot(x,y)

axes_arr[0][1].plot(x,y)
axes_arr[0][1].set_ylabel('Bla Bla')

axes_arr[1][0].plot(x,y)

axes_arr[1][1].plot(a,b)

fig.suptitle('Some Subplots')

plt.tight_layout()
fig.savefig('new_sublots.png', bbox_inches='tight')
```

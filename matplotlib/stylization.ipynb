```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,10,11)
```

```python
fig = plt.figure()

ax = fig.add_axes([0,0,1,1])

ax.plot(x,x, label='x, x');

ax.plot(x, x**2, label='x**2')

ax.legend()
```

```python
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
ax.plot(x,x, color = '#000000', label='X on X', lw=2, ls=":")
ax.plot(x,x+1, color = 'blue', label='X on X+1', lw=2, ls="-.", 
        marker='h', markersize=8, markerfacecolor='white', markeredgewidth=3, markeredgecolor = 'red')
```

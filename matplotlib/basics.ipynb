```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
x = np.arange(0,10)
```

```python
y = x*2
```

```python
plt.plot(x,y);
plt.title('Nigga');
plt.xlabel('distance');
plt.ylabel('time');
plt.xlim(0,6);
plt.ylim(0, 15);
plt.savefig('MyFirstPlot.png')
```
